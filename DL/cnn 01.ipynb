```python
# TensorFlow aur Keras import
import tensorflow as tf
from keras import Sequential
from keras.layers import (
    Input,
    Conv2D,
    MaxPooling2D,
    Flatten,
    Dense
)
```

```python

# CNN architecture create karna
model = Sequential([

    # Input image: 64 height, 64 width, 3 RGB channels
    Input(shape=(64, 64, 3)),

    # 16 filters, har filter ka size 3x3
    Conv2D(16, (3, 3), activation="relu"),

    # 2x2 window se feature map ka size reduce
    MaxPooling2D(pool_size=(2, 2)),

    # 32 filters se additional image patterns learn karna
    Conv2D(32, (3, 3), activation="relu"),

    # Feature map ka size phir reduce
    MaxPooling2D(pool_size=(2, 2)),

    # Feature maps ko one-dimensional vector banana
    Flatten(),

    # 64 neurons wali classification layer
    Dense(64, activation="relu"),

    # 3 classes: Apple, Banana, Orange
    Dense(3, activation="softmax")
])

# Model ko training ke liye configure karna
model.compile(
    optimizer="adam",  # Parameters update karne ka method
    loss="sparse_categorical_crossentropy",  # Integer class labels ke liye loss
    metrics=["accuracy"]  # Accuracy track karna
)

# Model ki architecture aur parameter count print karna
model.summary()
```

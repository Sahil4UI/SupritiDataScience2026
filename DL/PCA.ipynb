```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
data = {
    "Maths":[80,85,90,75,70,88,92,65,60,95],
    "Physics":[82,87,91,73,72,90,94,67,62,96],
    "Chemistry":[78,83,89,76,68,86,91,64,61,93],
    "English":[70,75,82,68,65,80,85,60,58,88],
    "Attendance":[90,92,95,85,80,93,96,75,70,98],
    "StudyHours":[6,7,8,5,4,7,9,3,2,10]
}
```

```python
df = pd.DataFrame(data)
```

```python
print(df)
```

```python
x = df[["Maths","Physics","Chemistry","English" ,"Attendance",  "StudyHours"]]
```

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
x_scaled
```

```python
pca = PCA(n_components=2)
```

```python
x_pca = pca.fit_transform(x_scaled)
```

```python
print(x_pca)
```

```python
# PCA KA RESULT => NEW DATA FRAME
pca_df = pd.DataFrame(
    x_pca,
    columns = ["Principal Component 1","Principal Component 2"]
)
```

```python
pca_df
```

```python
plt.figure(figsize=(8,8))
plt.scatter(
    pca_df["Principal Component 1"],pca_df["Principal Component 2"]
)
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("Students After PCA")
plt.show()
```

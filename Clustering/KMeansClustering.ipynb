```python
import pandas as pd
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv('xclara.csv')
```

```python
dataset.head()
```

```python
plt.figure(figsize=(12,6))
plt.scatter(dataset['V1'], dataset['V2'], c='b', s=30)
```

```python
kmeans = KMeans(n_clusters=3)
kmeans.fit(dataset)
```

```python
kmeans.cluster_centers_
```

```python
# Cluster labels
labels = kmeans.labels_

# Cluster centers
centers = kmeans.cluster_centers_

plt.figure(figsize=(10,6))

# Data points with different colors
plt.scatter(dataset['V1'], dataset['V2'],
            c=labels,
            s=40)

# Centroids
plt.scatter(centers[:,0], centers[:,1],
            c='red',
            marker='X',
            s=250,
            label='Centroids')

plt.title("K-Means Clustering")
plt.xlabel("V1")
plt.ylabel("V2")
plt.legend()
plt.grid(True)
plt.show()
```

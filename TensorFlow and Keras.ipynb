```python
import tensorflow as tf

#keras->sequential model
from keras import Sequential

# Dense is a fully connected network layer
from keras.layers import Dense

import numpy as np
```

```python
# Data Training
```

```python
# Har students ke 3 features
# 1. study hour
# 2. attendence
# 3. previous marks

x = np.array([[1,50,35],
             [2,55,40],
             [3,60,45],
             [4,70,55],
             [5,75,65],
             [6,85,70],
             [7,90,75],
             [8,95,80]], dtype=float)
```

```python
# target /output
y = np.array([0,0,0,1,1,1,1,1])
```

```python
# create model
model = Sequential([
    # hidden layer1
    Dense(8,
          activation='relu',
          input_shape=(3,)
          ),
    #hidden layer 2
    Dense(4,activation="relu"),


    # 1 neuron output : 0,1
    Dense(1,activation="sigmoid")
    
          ])
```

```python
# model compile
model.compile(
    # optimizer:model training k time apne parameter->update
    optimizer="adam",
    # loss
    # output - 0 or 1
    loss = "binary_crossentropy",
    # metrices:
    # training k time accuracy nikalni h
    metrics=["accuracy"]
)
```

```python
model.fit(x,
          y,
          epochs = 50,
          verbose=0
        #   verbose - training k output ->show
          )
```

```python
study_hour = 6
Attendance = 90
prev_marks = 75

newStudent = np.array([
    [study_hour,Attendance,prev_marks]
])
```

```python
prediction = model.predict(
    newStudent,
    verbose = 1
)
```

```python
print("probability:",prediction[0][0])
```

```python
# 0.5 se less - fail
# 0.5 se jyda - pass
if prediction[0][0] >= 0.5:
    print("pass")
else:
    print("fail")
```

```python
import numpy as np
import pandas as pd
```

```python
#Relu Function
# 1.ReLU - Rectified Linear Unit
# Neg Value -> 0
# Pos Value -> same

def relu(x):
    return np.maximum(0,x)
```

```python
values  = np.array([-5,-2,0,3,8])
```

```python
relu_output = relu(values)
```

```python
relu_output
```

```python
#SIGMOID - kissi bi value ko approx o aur 1 k bich convert krta h
```

```python
def sigmoid(x):
    return 1/(1+np.exp(-x))
```

```python
sigmoid_output = sigmoid(values)
```
